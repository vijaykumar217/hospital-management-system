```python
import gradio as gr

def generate_health_report(
    patient_name, age, gender, height, weight, blood_group,
    temperature_fahrenheit, systolic_bp, diastolic_bp, blood_sugar, oxygen, heart_rate, symptoms
):
    # Convert Fahrenheit to Celsius for internal logic
    temperature = (temperature_fahrenheit - 32) * 5/9

    # BMI ANALYSIS
    height_in_meters = height / 100
    bmi = weight / (height_in_meters ** 2)
    if bmi < 18.5:
        bmi_status = 'under weight!'
    elif bmi < 25:
        bmi_status = 'normal weight!'
    elif bmi < 30:
        bmi_status = 'over weight!'
    else:
        bmi_status = 'obese!'

    # AGE CATEGORY
    if age <= 12:
        age_category = 'child'
    elif age <= 19:
        age_category = 'teenager'
    elif age <= 59:
        age_category = 'adult'
    else:
        age_category = 'senior citizen!'

    # BODY TEMPERATURE ANALYSIS
    # Note: The original logic implicitly used Celsius-like thresholds, now `temperature` is in Celsius
    if temperature < 96.0:
        status = 'Low Body Tempetature!'
    elif temperature <= 97.5:
        status = 'Normal!'
    elif temperature <= 98.5:
        status = 'Mild Fever!'
    elif temperature <= 99.5:
        status = 'High Fever!'
    else:
        status = 'Very High Fever!'

    # BLOOD PRESSURE ANALYSIS
    if systolic_bp < 90 and diastolic_bp < 60:
        bp_status = 'low blood pressure!'
    elif systolic_bp < 120 and diastolic_bp < 80:
        bp_status = 'normal blood pressure!'
    elif systolic_bp < 129 and diastolic_bp < 80:
        bp_status = 'Elevated blood pressure!'
    elif systolic_bp < 139 or diastolic_bp < 90:
        bp_status = 'Stage 1 hypertension!'
    elif systolic_bp < 179 or diastolic_bp < 119:
        bp_status = 'Stage 2 hypertension!'
    else:
        bp_status = 'Hypertensive crisis!'

    # BLOOD SUGAR ANALYSIS
    if blood_sugar < 70:
        bs_status = 'low blood sugar!'
    elif blood_sugar < 99:
        bs_status = 'normal blood sugar!'
    elif blood_sugar < 125:
        bs_status = 'pre-diabetes!'
    elif blood_sugar < 199:
        bs_status = 'diabetes!'
    else:
        bs_status = 'severe diabetes!'

    # OXYGEN LEVEL ANALYSIS
    if oxygen < 90:
        oxygen_status = 'critical oxygen level!'
    elif oxygen < 95:
        oxygen_status = 'need observation!'
    else:
        oxygen_status = 'normal oxygen level!'

    # HEART RATE ANALYSIS
    if heart_rate < 60:
        hr_status = 'low heart rate!'
    elif heart_rate < 100:
        hr_status = 'normal heart rate!'
    else:
        hr_status = 'high heart rate!'

    # Decision Predictor
    symptoms = symptoms.lower()
    if 'fever' in symptoms and 'cough' in symptoms:
        disease = 'Viral Infection!'
    elif 'chest pain' in symptoms:
        disease = 'Heart Disease!'
    elif 'headache' in symptoms:
        disease = 'Migraine!'
    elif 'body pain' in symptoms:
        disease = 'Viral Fever!'
    elif 'stomach pain' in symptoms:
        disease = 'Gastric Problem!'
    elif 'sore throat' in symptoms:
        disease = 'Throat Infection!'
    else:
        disease = 'General Checkup Required!'

    # Doctor Recommendation
    if disease == "Heart Disease!":
        doctor = "Cardiologist"
    elif disease == "Migraine!":
        doctor = "Neurologist"
    elif disease == "Food Poisoning":
        doctor = "Gastroenterologist"
    elif disease == "Gastric Problem!":
        doctor = "Gastroenterologist"
    elif disease == "Throat Infection!":
        doctor = "ENT Specialist"
    elif disease == "Viral Infection!":
        doctor = "General Physician"
    elif disease == "Viral Fever!":
        doctor = "General Physician"
    else:
        doctor = "General Physician"

    # ADMISSION RECOMMENDATION
    if oxygen < 90:
        admission = "Emergency Treatment Required"
    elif status == "Very High Fever!":
        admission = "Emergency Treatment Required"
    elif disease == "Heart Disease!":
        admission = "Emergency Treatment Required"
    elif bp_status == "Hypertension Stage 2":
        admission = "Hospital Admission Recommended"
    elif bp_status == "Hypertensive Crisis!":
        admission = "Hospital Admission Recommended"
    elif bs_status == "Severe Diabetes!":
        admission = "Hospital Admission Recommended"
    else:
        admission = "Home Care"

    # MEDICINE RECOMMENDATION
    if disease == "Viral Infection!":
        medicine = "Paracetamol (Consult a doctor before use)."
    elif disease == "Heart Disease!":
        medicine = "Immediate consultation with a Cardiologist. Do not self-medicate."
    elif disease == "Migraine!":
        medicine = "Pain relief medicine as prescribed by a doctor."
    elif disease == "Viral Fever!":
        medicine = "Paracetamol and plenty of fluids."
    elif disease == "Gastric Problem!":
        medicine = "Antacid as advised by a doctor."
    elif disease == "Throat Infection!":
        medicine = "Warm salt-water gargling and prescribed medication."
    else:
        medicine = "No medicine recommendation. Please consult a General Physician."

    # DIET RECOMMENDATION
    if bs_status == "Severe Diabetes!":
        diet = "Strict diabetic diet with no sugary foods."
    elif bs_status == "Diabetes!":
        diet = "Low-sugar diet with whole grains and vegetables."
    elif bmi_status == "Obese!":
        diet = "Low-calorie diet with more fruits and vegetables."
    elif bmi_status == "Overweight!":
        diet = "Balanced diet with reduced fat and sugar."
    elif bmi_status == "Underweight!":
        diet = "High-protein and high-calorie nutritious diet."
    elif disease == "Gastric Problem!":
        diet = "Avoid spicy foods and eat easily digestible meals."
    elif disease == "Throat Infection!":
        diet = "Drink warm fluids and eat soft foods."
    else:
        diet = "Maintain a healthy balanced diet."

    # Format output as a comprehensive string
    report_str = f"FINAL HEALTH REPORT\n\n"
    report_str += f"Patient Name          : {patient_name}\n"
    report_str += f"Age                   : {age}\n"
    report_str += f"Gender                : {gender}\n\n"
    report_str += f"BMI                   : {round(bmi, 2)}\n"
    report_str += f"BMI Status            : {bmi_status}\n\n"
    report_str += f"Age Category          : {age_category}\n\n"
    report_str += f"Temperature Status    : {status}\n\n"
    report_str += f"Blood Pressure Status : {bp_status}\n\n"
    report_str += f"Blood Sugar Status    : {bs_status}\n\n"
    report_str += f"Oxygen Status         : {oxygen_status}\n\n"
    report_str += f"Heart Rate Status     : {hr_status}\n\n"
    report_str += f"Possible Disease      : {disease}\n\n"
    report_str += f"Recommended Doctor    : {doctor}\n\n"
    report_str += f"Admission Status      : {admission}\n\n"
    report_str += f"Diet Recommendation   : {diet}\n\n"
    report_str += f"Medicine Recommendation : {medicine}\n\n"
    report_str += f"      THANK YOU! STAY HEALTHY."

    return report_str
```

Now, let's create the Gradio interface using the `generate_health_report` function.

```python
inputs = [
    gr.Textbox(label="Patient Name"),
    gr.Slider(1, 120, value=30, label="Age", step=1),
    gr.Radio(["Male", "Female", "Other"], label="Gender"),
    gr.Slider(50, 250, value=170, label="Height (cm)", step=0.1),
    gr.Slider(20, 200, value=70, label="Weight (kg)", step=0.1),
    gr.Dropdown(["A+", "A-", "B+", "B-", "AB+", "AB-", "O+", "O-"], label="Blood Group"),
    gr.Slider(90.0, 110.0, value=98.6, label="Body Temperature (°F)", step=0.1), # Corrected range and label
    gr.Slider(70, 200, value=120, label="Systolic Blood Pressure"),
    gr.Slider(40, 150, value=80, label="Diastolic Blood Pressure"),
    gr.Slider(50.0, 300.0, value=100.0, label="Blood Sugar (mg/dL)", step=0.1),
    gr.Slider(70, 100, value=98, label="Oxygen Level (%)"),
    gr.Slider(40, 150, value=70, label="Heart Rate (BPM)"),
    gr.Textbox(label="Symptoms (comma-separated, e.g., fever, cough, headache)")
]

outputs = gr.Textbox(label="Health Report and Recommendations")

gradio_interface = gr.Interface(
    fn=generate_health_report,
    inputs=inputs,
    outputs=outputs,
    title="Hospital Management Decision System",
    description="Enter patient details to get a comprehensive health report and recommendations."
)

gradio_interface.launch(debug=True, share=True)
```
